# AND и OR: обучение одного нейрона

AND и OR — задачи бинарной классификации. Их таблицы истинности
линейно разделимы, поэтому для каждой достаточно одного нейрона с
сигмоидой. Сеть вычисляет `Z = X @ W + b`, применяет сигмоиду,
а затем обновляет веса по градиенту MSE с помощью Adam.

Сначала обучим отдельную модель для AND, затем другую модель для OR.
Сравним округлённые вероятности с ожидаемыми ответами.

```python
import numpy as np

from open_neuron.activations import sigmoid
from open_neuron.layers import Layer
from open_neuron.models import NeuralNetwork
from open_neuron.optimizers import Adam

np.set_printoptions(precision=3, suppress=True)
X = np.array([[0, 0], [0, 1], [1, 0], [1, 1]])
and_targets = np.array([[0], [0], [0], [1]])
or_targets = np.array([[0], [1], [1], [1]])

print('Входы:')
print(X)
print('Ответы AND:', and_targets.ravel())
print('Ответы OR: ', or_targets.ravel())
```

## Обучаем AND

`Layer(1)` создаёт один нейрон. Начальное состояние зафиксировано
seed, чтобы пример можно было повторить. `fit` выполняет прямой
проход, вычисляет градиенты и обновляет параметры на каждом пакете.
Здесь весь набор из четырёх примеров — один пакет.

```python
np.random.seed(7)
and_model = NeuralNetwork(
    [Layer(1, activation=sigmoid)],
    inputs_dim=X.shape[1],
    loss='mse',
    optimizer=Adam(learning_rate=1.0),
)
and_model.fit(X, and_targets, epochs=200, batch_size=4, shuffle=False)
and_predictions = and_model.predict(X)

print('Вероятности AND:')
print(and_predictions.ravel())
print('Классы AND:      ', np.round(and_predictions).astype(int).ravel())
print('Ожидалось:       ', and_targets.ravel())
```

## Обучаем OR

Для OR создаём новый экземпляр сети и новый оптимизатор, поэтому
параметры и состояние Adam не переносятся из обучения AND.

```python
np.random.seed(7)
or_model = NeuralNetwork(
    [Layer(1, activation=sigmoid)],
    inputs_dim=X.shape[1],
    loss='mse',
    optimizer=Adam(learning_rate=1.0),
)
or_model.fit(X, or_targets, epochs=200, batch_size=4, shuffle=False)
or_predictions = or_model.predict(X)

print('Вероятности OR:')
print(or_predictions.ravel())
print('Классы OR:      ', np.round(or_predictions).astype(int).ravel())
print('Ожидалось:      ', or_targets.ravel())
```
