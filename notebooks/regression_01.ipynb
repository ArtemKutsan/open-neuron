# Простая линейная регрессия

Обучим один полносвязный слой предсказывать числовую зависимость `y = 2x + 1`.
Для регрессии используем линейную активацию: выход не ограничивается
диапазоном вероятностей `[0, 1]`. Четыре примера пойдут в обучение,
а значение `x = 2` оставим для проверки после обучения.

```python
import numpy as np

from open_neuron.activations import linear
from open_neuron.layers import Layer
from open_neuron.losses import MSE
from open_neuron.models import NeuralNetwork
from open_neuron.optimizers import SGD

np.set_printoptions(precision=3, suppress=True)

X_train = np.array([[-2.0], [-1.0], [0.0], [1.0]])
y_train = 2 * X_train + 1
X_validation = np.array([[2.0]])
y_validation = 2 * X_validation + 1

print('Обучающие входы:')
print(X_train.ravel())
print('Целевые значения:')
print(y_train.ravel())
print('Проверочный вход и ожидаемый ответ:', X_validation.ravel(), y_validation.ravel())
```

## Создание и обучение модели

`Layer(1)` создаёт один нейрон с одним весом и смещением. Его выход
равен `a = x * weight + bias`, потому что активация линейная.
`batch_size=4` подаёт все обучающие примеры одним пакетом.
Проверочный пример не участвует в обновлении параметров.

```python
np.random.seed(7)
model = NeuralNetwork(
    [Layer(1, activation=linear)],
    inputs_dim=X_train.shape[1],
    loss='mse',
    optimizer=SGD(learning_rate=0.1),
)

initial_prediction = model.predict(X_validation)
initial_loss = MSE().evaluate_loss(y_validation, initial_prediction)
print('Начальное предсказание для x=2:', initial_prediction.ravel())
print(f'Начальная MSE на проверочном примере: {initial_loss:.6f}')

model.fit(
    X_train,
    y_train,
    epochs=100,
    batch_size=4,
    shuffle=False,
    validation_data=(X_validation, y_validation),
)

prediction = model.predict(X_validation)
final_loss = MSE().evaluate_loss(y_validation, prediction)
print('Предсказание для x=2:', prediction.ravel())
print('Ожидаемое значение:', y_validation.ravel())
print(f'Итоговая MSE на проверочном примере: {final_loss:.8f}')
```
